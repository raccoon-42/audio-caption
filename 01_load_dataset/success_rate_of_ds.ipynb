```python
from datasets import load_from_disk
import os
from tqdm import tqdm

ds = load_from_disk("../musiccaps/dataset_audio")

count = 0
for i in tqdm(range(len(ds))):
    row = ds._data.slice(i, 1)
    ytid = row["ytid"][0].as_py()
    start = row["start_s"][0].as_py()
    end = row["end_s"][0].as_py()
    path = f"../musiccaps/segments/{ytid}_{start:.2f}_{end:.2f}.wav"
    
    if not os.path.exists(path):
        count += 1

print(f"Number of examples with no audio: {count}")
print(f"Success rate: {(len(ds) - count) / len(ds) * 100:.2f}%")
```
